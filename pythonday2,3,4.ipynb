##  nested if 

an if condition can besaid when it is a part of another if block of code. the nested if will only be tested if the parent'if'expression happens to be true 

```python
i=25
if(i>=10):# the parent if condition 
    # these expression are checked only if the parent if condition is true 
    if(i<15):  # nested if condition 
        print("i is smaller than 15")
    if (i<12):
        print("i is smaller than 12")
else:
    print("i is greater than 15")
```

```python
i=int(input("enter the number"))
if(i>=100):
    print("blood pressure is above 100")# the parent if condition 
    # these expression are checked only if the parent if condition is true 
if(i<120):  # nested if condition 
        print("normal range")
elif 120 <= i <=129:
        print("elivated range")
else:
    print("high blood pressure")
```
